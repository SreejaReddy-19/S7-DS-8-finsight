```python
import matplotlib.pyplot as plt
y1 =[] #stores +ve values
y2 =[] #stores -ve values
x = range(-100,100,10) #Generous values from -100 to 90 with an interval of 10(-100,-90,-80...)
for i in x: y1.append(i**2) #y=x^2
for i in x: y2.append(-i**2)
plt.plot(x,y1)
plt.plot(x,y2)
plt.xlabel("x")
plt.ylabel("y")
plt.ylim(-2000,2000) #yaxis limit
plt.axhline(0) #horizontal line
plt.axvline(0) #vertical line
plt.savefig("quad.png")
plt.show()
```
